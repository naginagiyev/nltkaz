```python
import pandas as pd 

path = "C:\\Users\\nagin\\Downloads\\azspeech_sentences.csv"
df = pd.read_csv(path)
df
```

```python
from tqdm import tqdm

all_words = []
sentences = df['Sentences'].values.tolist()

for sentence in tqdm(sentences, desc="COUNTING"):
    all_words.extend(sentence.split())
```

```python
print(len(set(all_words)))
```
